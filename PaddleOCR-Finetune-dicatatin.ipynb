```python
from paddleocr import PaddleOCR
import time

print("--- Memuat Otak Bawaan PaddleOCR ---")
# Kita gunakan lang='en' karena struktur model Inggrisnya jauh lebih tangguh 
# untuk membaca huruf Latin bercampur angka ketimbang model 'id' (Indonesia).
ocr_engine = PaddleOCR(use_angle_cls=True, lang='en', show_log=False, use_gpu=False)

# Ganti dengan path foto catatanmu
image_path = 'test1.jpg' 

print(f"\nMengekstraksi teks dari: {image_path}...\n")
start_time = time.time()

# Eksekusi OCR
result = ocr_engine.ocr(image_path, cls=True)

if not result or result[0] is None:
    print("Mata OCR buta. Tidak ada teks yang terdeteksi.")
else:
    print("================ HASIL EKSTRAKSI ================\n")
    
    total_conf = 0
    lines_count = 0
    
    for line in result[0]:
        # line[0] adalah koordinat kotak (bounding box)
        # line[1][0] adalah teks yang dibaca
        # line[1][1] adalah confidence score (0.00 - 1.00)
        
        text = line[1][0]
        confidence = line[1][1]
        
        # Cetak dengan format [Score] Teks
        print(f"[{confidence:.2f}] {text}")
        
        total_conf += confidence
        lines_count += 1
        
    print("\n================ STATISTIK ================")
    avg_conf = total_conf / lines_count if lines_count > 0 else 0
    print(f"Rata-rata Confidence Score: {avg_conf:.2f}")
    print(f"Waktu Proses CPU          : {time.time() - start_time:.2f} detik")
```
